## Different Lenses

Any lens that is thicker at its center than at the edges is a **converging** lens with a positive focal point, i.e. $f > 0$.
Any lens that is thicker at its *edges* than at its *center* is a **diverging** lens with a negative focal point$f < 0$.

## Lensmaker's equation (for a thin lens)

$$
\frac{1}{f} = (n - 1)(\frac{1}{R_1} - \frac{1}{R_2})
$$

Where $R_1 > 0$ and $R_2 < 0$ when $C_1$ is on the same side as the outgoing light. In this case, the center of curvature is on the opposite side, then $R_1$ is positive.

(This tells you why the lens is converging).

### Ray 1 (purple)

This is the parallel incident wave which refracts to pass through the second ofcal point $F_2$. From the image, from point $Q$, it will leave that point. It will then travel parallel to the axis of the purple line. As it travels in a straight, horizontal line. But when it reaches the lens, it **refracts**. At that point it bends and as it bends it will cross the second focal point $F_2$.

### Ray 2 (green)

This leaves point $Q$ and goes straight until it reaches the lens. The surface is almost parallel, so the ray becomes a straight line with no bending (i.e. it does not change direction). 

### Ray 3 (oragne)

When it reaches the lens and passes through it, it travels parallel to the x-axis. This is the inverse of ray 1.

Same as before, we have three (3) rays again.

### Ray 1 (pruple)

Parallel incident ray which leaves its point $Q$ and travels parallel to the x-axis. Upon reaching the lens, it bends *upward* and travels far away. Tracing the line babackward passes through the second focal l point $F_2$. The ray seems to come from $F_2$.

### Ray 3 (orange)

Aimed at the first focal point $F_1$ and emerges parallel to the x-axis because the lens is in the way.

### Ray 2 (green)

Ray through the center of the lens does not deviate *appreciably*. 

### Dashed lines point of intersection

This is where the image is formed. The image is on the same side as the object, but it is a **virtual image**. It's pointing up, and the same orientation as the original image. The formed image is *smaller* than the original object

## Defects of vision

Normal eye are relaxed when it looks at a far-away object. In this state the muscle is relaxed and the lens is at its flattest.

In a myopic eye (i.e. **near-sighted**), the eyeball is too long from front to back in comparison with the radius of curvature of the cornea. See the lower diagram in the slides. The eyeball is "stretched" compared to a normal one. This bends the light too strongly, resulting in the black dot (the ???) before it reaches the eye.

## Nearsighted Correction (myopia)

Beyond 50 centimeters, all perceived objects are blurry. Since the eye can see clearly up to that distance, glasses create virtual images for distant objects at 50 cm. 

## Farsighted Correction (hyperopic)

The spreading rays are very hard to bend and focus. In a farsighted eye, they cannot be bent, so the rays meet behind the retina. The resulting image is **not** focused on the retina. This can be caused if the eyeball is too short, or the lens is too weak.
