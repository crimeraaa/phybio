15. A wave on a string is described by the following function:

$$
y(x, t) = (0.0050 \; \text{m})
    \cos((4.0 \; \dfrac{\text{rad}}{\text{m}}) x
        - (12 \; \dfrac{\text{rad}}{\text{s}}) t
        )
$$

What are the wave speed $v$ and direction of travel?

The above equation describes a sinusoidal wave function moving in the $+x$ direction. The general form is as follows, with values annotated:

$$
\begin{align}
y(x, y) &= A \cos(k x - \omega t) \\
\text{where:} \\
A &= 0.0050 \; \text{m} \\
k &= 4.0 \; \frac{\text{rad}}{\text{m}} \\
x &= \; ? \; \text{m} \\
\omega &= 12 \; \frac{\text{rad}}{\text{s}} \\
t &= \; ? \; \text{s} \\
\end{align}
$$

Given the angular frequency $\omega$, we can solve for the frequency $f$.

$$
\begin{align}
\omega &= 2 \pi f \\
f &= \frac{\omega}{2 \pi} \\
  &= \frac{12 \; \dfrac{\text{rad}}{\text{s}}
          }
          {2 \pi} \\
  &= 1.91 \; \frac{1}{\text{s}} \\
  &= 1.91 \; \text{Hz}
\end{align}
$$

We can also solve for the wavelength $\lambda$ using the formula for the wave number $k$.

$$
\begin{align}
k &= \frac{2 \pi}{\lambda} \\
\lambda k &= 2 \pi \\
\lambda &= \frac{2 \pi}{k} \\
        &= \frac{2 \pi}{4.0 \; \dfrac{\text{rad}}{\text{m}}} \\
        &= 1.57 \; \text{m}
\end{align}
$$

Finally, we can use the wavelength $\lambda$ and frequency $f$ to solve for the wave speed $v$.

$$
\begin{align}
v &= \lambda f \\
  &= 1.57 \; \text{m} \times 1.91 \; \frac{1}{\text{s}} \\
  &= 3.00 \; \frac{\text{m}}{\text{s}}
\end{align}
$$

16. A sinusoidal wave on a string has a frequency $f = 50 \; \text{Hz}$ and travels at $v = 20 \; \frac{\text{m}}{\text{s}}$. What is the wave number $k$?

In order to determine the wave number $k$, we first need the wave length $\lambda$:

$$
\begin{align}
v &= \lambda f \\
\lambda &= \frac{v}{f} \\
\lambda &= \frac{20 \; \dfrac{\text{m}}{\text{s}} }
                {50 \; \dfrac{1}{\text{s}}} \\
        &= 0.4 \; \text{m}
\end{align}
$$

Now we can determine the wave number.

$$
\begin{align}
k &= \frac{2 \pi}{\lambda} \\
  &= \frac{2 \pi}{0.4 \; \text{m}} \\
  &= 15.7 \; \frac{\text{rad}}{\text{m}}
\end{align}
$$

17. A string has a mass $m = 5.0 \; \text{g}$ and a length $L = 2.0 \; \text{m}$, and is under a tension $F = 80 \; \text{N}$. What is the speed of transverse waves $v$ on this string?

We need to calculate the linear mass density $\mu$, which is the just the mass $m$, in kilograms, per unit length $L$, in meters.

$$
\begin{align}
\mu &= \frac{m}{L} \\
    &= \frac{5.0 \; \text{g} \times \dfrac{1 \; \text{kg} }{1000 \; \text{g}} }
            {2.0 \; \text{m}} \\
    &= \frac{0.005 \; \text{kg}}{2.0 \; \text{m}} \\
    &= 0.0025 \; \frac{\text{kg}}{\text{m}}
\end{align}
$$

Thus the speed of transverse waves $v$ can be determined as follows:

$$
\begin{align}
v &= \sqrt{\frac{F}{\mu}} \\
  &= \sqrt{\frac{80 \; \text{N}}
                {0.0025 \; \frac{\text{kg}}{\text{m}} }
          } \\
  &= \sqrt{\frac{80 \; \text{kg} \cdot \text{m} \cdot \text{s}^{-2}}
                {0.0025 \; \frac{\text{kg}}{\text{m}} }
          } \\
  &= \sqrt{\frac{80 \; \text{kg} \cdot \text{m} \cdot \text{s}^{-2}}
                {0.0025}
           \frac{\text{m}}{\text{kg}}
          } \\
  &= \sqrt{\frac{80}{0.0025} \text{m}^2 \cdot \text{s}^{-2}} \\
  &= 179 \; \frac{\text{m}}{\text{s}}
\end{align}
$$

18. The tension $F$ in a string is increased by a factor of 9 and, at the same time, its linear mass density $\mu$ is increased by a factor of 4. BY what factor does the wave speed change?

19. On the same string with fixed tension, a wave's amplitude $A$ is tripled and its frequency $f$ is halved. By what factor does the average power $P_{av}$ change?

$$
\begin{align}
P_{av} &= \frac{1}{2} \sqrt{\mu F} \times \omega^2 \times 3 A^2 \\
\text{since:} \\
\omega &= 2 \pi f \\
\text{thus:} \\
P_{av} &= \frac{1}{2} \sqrt{\mu F} \times (2 \pi  f)^2 \times 3A^2 \\
P_{av} &= \frac{1}{2} \sqrt{\mu F} \times (2 \pi \dfrac{1}{2} f)^2 \times 9A \\
P_{av} &= \frac{1}{2} \sqrt{\mu F} \times \dfrac{1}{2}(2 \pi f)^2 \times 9A \\
P_{av} &= \frac{1}{2} \sqrt{\mu F} \times \dfrac{1}{4}(2 \pi f) \times 9A \\
\end{align}
$$

Thus the factor is $\dfrac{1}{4} \times 9 = 2.25$.

23. A guitar string has a vibrating length $L = 0.65 \; \text{m}$ and a linear mass density $\mu = 4.0 \dfrac{\text{g}}{\text{m}}$. What tension $F$ is required for a fundamental frequency $f_1 = 110 \; \text{Hz}$?

$$
\begin{align}
f_1 &= \frac{1}{2 L} \sqrt{\frac{F}{\mu}} \\
( f_1 &= \frac{1}{2 L} \frac{\sqrt{F}}{\sqrt{\mu}} ) \frac{2 L}{1} \\
( 2 L f_1 &= \frac{\sqrt{F}}{\sqrt{\mu}} ) \frac{\sqrt{\mu}}{1} \\
( \sqrt{F} &= 2 L f_1 \sqrt{\mu} )^2 \\
  F &= 4 L^2 {f_1}^2 \mu \\
    &= 4 \times (0.65 \; \text{m})^2
       \times (110 \; \dfrac{1}{\text{s}})^2
       \times 4.0 \; \dfrac{\text{g}}{\text{m}}
       (\dfrac{1 \; \text{kg}}{1000 \; \text{g}}) \\
    &= 82 \; \text{kg} \cdot \text{m} \cdot \text{s}^{-2} \\
    &= 82 \; \text{N}
\end{align}
$$
