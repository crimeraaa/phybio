14. A sound wave in air has a pressure amplitude $p_{max} = 3.0 \; \text{Pa}$. Use $\rho = 1.20 \; \dfrac{\text{kg}}{\text{m}^3}$ and wave speed $v = 344 \; \dfrac{\text{m}}{\text{s}}$. What is the intensty of the wave?

$$
\begin{align}
I &= \frac{{p_{max}}^2}
          {2 \rho v} \\
  &= \frac{(3.0 \; \text{Pa})^2}
          {2
           \times 1.20 \; \dfrac{\text{kg}}{\text{m}^2}
           \times  344 \; \dfrac{\text{m}}{\text{s}}
          } \\
  &= \frac{9.0 \; \text{Pa}^2}
          {825.6 \; \text{kg} \cdot \text{m}^{-1} \cdot \text{s}^{-1}} \\
  &= 0.0109 \; \frac{\text{W}}{\text{m}^2} \\
  &= 1.09 \times 10^{-2} \; \frac{\text{W}}{\text{m}^2} \\
\end{align}
$$

15. For the same wave as above, what is the sound intensity level?

$$
\begin{align}
\beta &= (10 \; \text{dB}) \log_{10}(\frac{I}{I_0}) \\
      &= (10 \; \text{dB}) \log_{10}(\frac{1.09 \times 10^{-2} \; \dfrac{\text{W}}{\text{m}^2} }
                                          {10^{-12} \; \dfrac{\text{W}}{\text{m}^2} }
         ) \\
      &= 10 \; \text{dB} \times 10 \\
      &= 100 \; \text{dB}
\end{align}
$$

```python
p_max = 3.0  ## Pa
rho   = 1.20 ## kg / m^3
v     = 344.0 ## m / s
I     = p_max**2 / (2 * rho * v) ## W / m^2
display(I)

I0 = 10**-12 ## W / m^2
beta = 10 * math.log10(I / I0)
display(math.log10(I / I0))
display(beta)
```

18. An organ pipe has a length $L = 0.60 \; \text{m}$. Use wave speed $v = 344 \; \dfrac{\text{m}}{\text{s}}$. What are the fundamental frequencies if the pipe is (i) open at both ends, and (ii) closed at one end?

For an open pipe:

$$
\begin{align}
f_1 &= \frac{v}{2 L} \\
    &= \frac{344 \; \cfrac{\text{m}}
                         {\text{s}}
            }
            {2 \times 0.60 \; \text{m}} \\
    &= 287 \; \frac{1}{\text{Hz}} \\
    &= 287 \; \text{Hz}
\end{align}
$$

For a closed pipe:

$$
\begin{align}
f_1 &= \frac{v}{4 L} \\
     &= \frac{344 \; \cfrac{\text{m}}
                         {\text{s}}
            }
            {4 \times 0.60 \; \text{m}} \\
\end{align}
$$

```python
L = 0.60 ## m
v = 344.0 ## m / s
display(L, v)

f_1 = v / (2 * L)
display(f_1)

f_1 = v / (4 * L)
display(f_1)
```

19. A pipe closed at one end has a fundamental frequency $f_1 = 220 \; \text{Hz}$ with wave speed $v = 344 \; \frac{\text{m}}{\text{s}}$. What is its length, and which of $440 \; \text{Hz}$ and $660 \; \text{Hz}$ can it produce?

$$
\begin{align}
f_1 &= \frac{v}{4 L} \\
4 L f_1 &= v \\
L &= \frac{v}{4 f_1} \\
  &= \frac{344 \; \cfrac{\text{m}}{\text{s}} }
          {4 \times 220 \; \cfrac{1}{\text{s}} } \\
  &= 0.391 \; \text{m}
\end{align}
$$

```python
f_1 = 220.0 ## Hz
v   = 344.0 ## m / s
display(v, f_1)

L = v / (4 * f_1)
display(L)
display(344 / (4 * 220))

for i in range(1, 10, 2):
    display(f_1 * i)
```

23. A listener walks at $v_L = 20 \; \dfrac{\text{m}}{\text{s}}$ toward a stationary source (i.e. $v_S = 0 \; \dfrac{\text{m}}{\text{s}})$ that emits $f_S = 500 \; \text{Hz}$ with a wave speed $v = 344 \; \dfrac{\text{m}}{\text{s}}$. What frequency does the listener hear?

$$
\begin{align}
f_L &= \frac{v + v_L}{v + v_S} f_S \\
    &= \frac{344 \; \cfrac{\text{m}}{\text{s}} + 20 \; \cfrac{\text{m}}{\text{s}}}
            {344 \; \cfrac{\text{m}}{\text{s}} + 0  \; \cfrac{\text{m}}{\text{s}}}
       \times 500 \; \text{Hz} \\
    &= \frac{364 \; \cfrac{\text{m}}{\text{s}} }
            {344 \; \cfrac{\text{m}}{\text{s}} }
      \times 500 \; \text{Hz}
\end{align}
$$

```python
v_L = 20.0
v_S = 0.0
f_S = 500.0
v   = 344.0
display(v_L, v_S, f_S, v)

f_L = (v + v_L) / (v + v_S) * f_S
display(f_L)
```
