## Sound Waves

Sound is a **longitudinal wave** in a medium, i.e. the medium moves back and forth along the same line the wave is travelling (i.e. the particles are **parallel** to the wave travel).

RECALL: Contrast it to a **transverse wave**, where the medium moves perpendicular to the direction that the wave travels. If the wave travels in a direct manner, since the particles are perpendicular to the motion thereof, the particles will be moving *up and down*.

Humans' **audible**range is between about $20 \text{ Hz}$ and $20,000 \text{ Hz}$. Below $20 \text{ Hz}$ is so-called **infrasound** (i.e. the spectrum used by elephants whales to communicate). Above $20,000 \text{ Hz}$ is **ultrasound**, e.g. how bats navigate, and 3D imaging of babies.

Mathematically, we write a **sinusoidal wave** travelling in the positive x-direction. It is exactly the way we wrote **transverse waves** from the last chapter. The folowing function is read as: "y of x and t is the displacement of a particle sitting at a position x and time t".

$$
y(x, t) = A \cos(k x - \omega t)
$$

Where:

1. $y(x, t)$ is the displacement of the particle sitting at position $x$ at time $t$.
2. $A$ is the displacement amplitude, i.e. how far the particle swings from its resting spot.
3. $k$ is the wave number, also equivalenn to $\frac{2 \pi}{\lambda}$.

Note that $y$ is not *height*, but rather a displacement along the direction of travel. 

## Different Waves to Describe a Sound Wave

### 1 of 3

If you are going to look at $y > 0$, this is whereever the curve is above the x-axis. The particle  that sits at that $x$ has been pushed to the **right** of where it normally sits. Then, for $y < 0$, the particle at this particular $x$ has been pushed to the **left**.

IMPORTANT: The row of dots at the bottom and the open circles are where each particle would sit if nothing were disturbing it (i.e. evenly spaced, at rest). The **filled dots** are where the particles are at "right now", the little arrows show which way they are moving (i.e. direction.)

The **vertical dashed lines** indicate where $y$ shifts from $y > 0$ (+) to $y < 0$ (-) and vice versa. The **dashed line** itself represents the convergence point.

Right under the line, the particels have "piled up" close together- i.e. **compression**.

The crossing, the **next dashed line**, is the oppoite- i.e. $y$ shifts from $y < 0$ (-) to $y > 0$ (+). In other words, particles on both sides are moving away from this point. The particles below have spread out (i.e. **rerfaction**). 

The graph overall represents where the medium is **getting denser** (compression) and **getting thinner** (rerfaction).

### 3 of 3: Describing Sound Mathematically as a displacement wave 

Recall the formula for a sound wave propagating in the positive x-direction: $y(x, y) = A \cos(k x - \omega t)$. The same sound wave can alternatively be described mathematicaly as a pressure wave:

$$
p(x, t) = B k A \sin(k x - \omega t)
$$

Where $B$ is the **bulk modulus**, it measures how strongly the medium resists being compressed. Note that $\sin$ is used rather than $\cos$, hence the $90 \degree$ phase shift.

The quantity $B k A$ is often called the Pressure amplitude $P_{max}$. Remember that the wave number $k = \frac{2 \pi}{\lambda}$.

Why $\sin$? The pressure fluctutation comes from how tightly packed the particles are locally, which depends on how fast the displacement is changing with the position, NOT the displacement itself. The rate of change is a derivative. The derivative of $\cos$ is $\sin$.

## Speed of Sound Waves

The speed of sound depends on the characteristics of the medium, mainly how stiff the medium is against compression.

In a fluid like water, the speed of sound $v$ is:

$$
v = \sqrt{\frac{B}{\rho}}
$$

where $B$ is the Bulk modulus of the fluid and $\rho$ the density of the fluid.

In a solid rod or bar, the speed of sound $v$ is:

$$
v = \sqrt{\frac{Y}{\rho}}
$$
where $Y$ is the Young's modulus of the rod materials and $\rho$ is the density of the rod materials.

In an ideal gas, such as air, the speed of sound $v$ is:

$$
v = \sqrt{\frac{\gamma R T}{M}}
$$

where $\gamma$ is the ratio of heat capacitiies,
$R$ is the gas constant,
$T$ is the absolute temperature (i.e. in Kelvin $K$),
and $M$ is the molar mass of the gass.

## Table 16.1: Speed of Sound in Various Bulk Materials

In gases, lighter gases generaly carry sound faster. E.g. hydrogen is faster than helium, which inturn is faster than air.

In liquids, higher temperatures generaly carry sound faster.

In solids, aluminum is faster than lead than steel.

Overall, solid beats liquid, liquid beats gases. Solids have the strongest restoring force between neighboring atoms, relative to their air density. So a disturbance "shoots through them" fastest.

## Sound Intensity

We know that sound waves carry energy, which is why we can feel loud bass notes in our chests at concerts.

The **wave intensity** $I$ is defined as: *"the time average rate at which wave energy is transported per unit area across a surface perpendicular to the direction of propagation"*. The **intensity** of a **sinusoidal sound wave** is proportional to the square of the pressure amplitude.

$$
I = \frac{{p_{max}}^2}{2 \rho v} = \frac{{p_{max}}^2}{2 \sqrt{\rho B}}
$$

Where:

1. $I$ is the **intensity** of a **sinusoidal sound wave** in fluid.
2. $p_{max}$ is the **pressure amplitude**.
3. $\rho$ is the **density** of the fluid.
4. $v$ is the wave speed.
5. $B$ is the **bulk modulus** of the fluid.

## The Decibel Scale

The human ear is sensitive over a broad range of intensities. A logarithmic measure of intensity, **sound intensity level**, expresed in **decibels** (dB), is often used:

$$
\beta = (10 \; \text{dB}) \log \frac{I}{I_{0}}
$$

Where:

1. $\beta$ is the **sound intensity level**.
2. $\log$ is the logarithm to base-10.
3. $I$ is the **intensity** of the sund wave.
4. $I_{0}$ is the **reference intensity**, i.e. $I_{0} = 10^{-12} \; \frac{\text{W}}{\text{m}^2}$.

The chosen reference intensity $I_{0}$ is approximately the threshold of human hearing, at $1000 \; \text{Hz}$.

1. (a) If two (2) sounds differ by $5.00 \; \text{dB}$, find the ratio of the intensity of the louder sound to that of the softer one.


First, let's recall the equation for the **decibel scale**, which is a logarithmic measure of **sound intensity level**:

$$
\beta = (10 \; \text{dB}) \log_{10}\frac{I}{I_0}
$$

where:
- $\beta$ is the **sound intensity level**, in **decibels** ($\text{dB}$).
- $\log_{10}$ is the logarithm to base-10.
- $I$ is the **sound wave**, in watts-per-square-meter ($\frac{\text{W}}{\text{m}^2}$).
- $I_0$ is the **reference intensity** which is approximately the threshold of human hearing at $1000 \; \text{Hz}$, the value $10^{-12} \frac{\text{W}}{\text{m}^2}$.

So we need to solve for the **sound wave intensity** (not the sound intensity level!) for both sounds. Let's derive the formula for $I$:

$$
\begin{align}
[ \beta &= (10 \; \text{dB}) \log_{10}\frac{I}{I_0} ] \frac{1}{10 \; \text{dB}} \\
  \frac{\beta}{10 \; \text{dB}} &= \log_{10}\frac{I}{I_0}\\
\end{align}
$$

We've hit a minor roadblock: we can't isolate $I$ without converting the logarithm back to its exponent form. Let's recall the definition of a logarithm:

$$
\begin{align}
b^y = x \\
\log_b{x} = y
\end{align}
$$

where
- $b$ is our base, 10.
- $x$ is the value applied to the logarithm function, in our case $\frac{I}{I_0}$.
- $y$ is the result of our logarithm function, in our case $\frac{\beta}{10 \; \text{dB}}$.

So we can rewrite the previous equation:

$$
\begin{align}
\log_{10} \frac{I}{I_0} &= \frac{\beta}{10 \; \text{dB}} \\
( 10^{\beta / 10 \; \text{dB}} &= \frac{I}{I_0} ) \frac{I_0}{1} \\
I_0 10^{\beta / 10 \; \text{dB}} &= I \\
I &= I_0 10^{\beta / 10 \; \text{dB}}
\end{align}
$$

We can now get the ratio for some $I_{loud}$ and $I_{soft}$. We need to assign arbitrary sound intensity levels- let's try $\beta_{loud} = 140 \; \text{dB}$ and $\beta_{soft} = 135 \; \text{dB}$.

$$
\begin{align}
I_{loud} &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{\beta_{loud} / {10 \; \text{dB}}} \\
         &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{{140 \; \text{dB}} / {10 \; \text{dB}}} \\
         &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{14} \\
         &= 100 \; \frac{\text{W}}{\text{m}^2} \\
\\
I_{soft} &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{\beta_{soft} / {10 \; \text{dB}}} \\
         &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{{135 \; \text{dB}} / {10 \; \text{dB}}} \\
         &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{13.5} \\
         &= 31.62278 \; \frac{\text{W}}{\text{m}^2} \\
\\
\frac{I_{loud}}{I_{soft}}
    &= \frac{100 \; \frac{\text{W}}{\text{m}^2} }
            {31.62278 \; \frac{\text{W}}{\text{m}^2} } \\
    &= 3.162278
\end{align}
$$

Let's confirm that the ratio holds true for other values as long as $I_{soft} = I_{loud} - 5 \; \text{dB}$. Let's try $\beta_{loud} = 120 \; \text{dB}$ and ${\beta_{soft} = 115 \; \text{dB}}$.

$$
\begin{align}
I_{loud} &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^\frac{120 \; \text{dB}}{10 \; \text{dB}} \\
         &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^\frac{120}{10} \\
         &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{12} \\
         &= 1 \; \frac{\text{W}}{\text{m}^2}
\\
I_{soft} &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^\frac{115 \; \text{dB}}{10 \; \text{dB}} \\

         &= (10^{-12} \; \frac{\text{W}}{\text{m}^2}) \; 10^{11.5} \\
         &= 0.3162278 \; \frac{\text{W}}{\text{m}^2} \\
\frac{I_{loud}}{I_{soft}}
    &= \frac{1 \; \frac{\text{W}}{\text{m}^2}}
            {0.3162278 \; \frac{\text{W}}{\text{m}^2} } \\
    &= 3.162278
\end{align}
$$

We have the same ratio.

## 16.41: Plane

On the planet Arrakis, a male ornithoid is flying towards his mate at $v = 25.0 \; \frac{\text{m}}{\text{s}}$ while singing at a frequency $f = 1200 \; \text{Hz}$. If the stationay female hears a tone of $f = 1240 \; \text{Hz}$, what is the speed of sound in the atmosphere of Arrakis?

Solving this requires us to review the **Doppler effect**, which is when a shift in sound frequency occurs due to motion of said sound's source $L$ in relation to the listener $L$ (or both):

$$
f_1 = \frac{v + v_1}{v + v_s} f_s
$$

Where:

1. $f_1$ is the frequency heard by the listener $L$.
2. $v$ is the speed of the sound.
3. $v_1$ is the velocity of the listener $L$ in relation to the sound source $S$. If $v_1 > 0$ then $L$ is moving towards $S$, else if $v_1 < 0$ then $L$ is moving away.
4. $v_s$ is the velocity of the source. Follows similar rules regarding signedness as $v_1$.
5. $f_s$ is the frequency of the sound emitted by the source.

Given:

1. $S$, the source is the male ornithoid.
2. $L$, the listener, is the female ornihoid.
3. $f_s = 1200 \; \text{Hz}$ is the frequency that $S$ is singing at.
4. $v_s = 25.0 \; \frac{\text{m}}{\text{s}}$ is the velocity of $S$ flying towards $L$.
5. $f_1 = 1240 \; \text{Hz}$ is the frequency that $L$ actually hears.
6. $v_1 = 0.00 \; \frac{\text{m}}{\text{s}}$ because $L$ is stationary.

Using the above equation, we are missing the main speed of sound $v$. Let's isolate it first:

$$
\begin{align}
                      [ f_1       &= \frac{v + v_1}{v + v_s} f_s ] \frac{1}{f_s} \\
            [ \frac{f_1}{f_s}     &= \frac{v + v_1}{v + v_s} ] \frac{v + v_s}{1} \\
({v + v_s}) \frac{f_1}{f_s}       &= v + v_1 \\
\frac{f_1 v + f_1 v_s}{f_s}       &= v + v_1 \\
\frac{f_1 v + f_1 v_s}{f_s} - (v) &= v + v_1 - (v) \\
\frac{f_1 v + f_1 v_s}{f_s} -  v  &= v_1 \\
\frac{f_1 v + f_1 v_s}{f_s} - \frac{f_s v}{f_s}  &= v_1 \\
\frac{f_1 v + f_1 v_s - f_s v}{f_s} &= v_1 \\
\frac{f_1 v - f_s v + f_1 v_s}{f_s} &= v_1 \\
\frac{v \times (f_1 - f_s) + f_1 v_s}{f_s} &= v_1 \\
\frac{v \times (f_1 - f_s) + f_1 v_s}{f_s} (\frac{f_s}{1}) &= v_1 (\frac{f_s}{1}) \\
v \times (f_1 - f_s) + f_1 v_s &= f_s v_1\\
v \times (f_1 - f_s) + f_1 v_s - (f_1 v_s) &= f_s v_1 - (f_1 v_s) \\
v \times (f_1 - f_s) &= f_s v_1 - f_1 v_s \\
\frac{v \times (f_1 - f_s)}{f_1 - f_s} &= \frac{f_s v_1 - f_1 v_s}{f_1 - f_s} \\
v &= \frac{f_s v_1 - f_1 v_s}{f_1 - f_s}
\end{align}
$$

After all that, we can finally solve for $v$:

$$
\begin{align}
v &= \frac{f_s v_1 - f_1 v_s}
          {f_1 - f_s} \\
  &= \frac{ 1200 \; \text{Hz} \times 0.00 \frac{\text{m}}{\text{s}} - 1240 \; \text{Hz} \times 25.0 \; \frac{\text{m}}{\text{s}}}
          { 1240 \; \text{Hz} - 1200 \; \text{Hz}} \text{ and Hz} = \frac{1}{\text{s}} \\
  &= \frac{ 0.00 - 31000 \frac{\text{m}}{\text{s}^2} }
          { 40 \; \frac{1}{\text{s}} } \\
  &= -775 \frac{\text{m}}{\text{s}}
\end{align}
$$
