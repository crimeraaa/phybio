15.10. A water wave travelling in a straight line on a lake is described by the equation:

$$
y(x, t) = (2.75\;\mathrm{cm})
    \times \cos(0.410 \mathrm{\frac{rad}{cm}} x
                + 6.20 \mathrm{\frac{rad}{s}} t)
$$

Where $y$ is the displacement perpendicular to the undisturbed surface of the lake.

This is an example of a *Sinusoidal wave* function, given by the following equation (15.7):

$$
y(x, t) = A \cos(k t - \omega t)
$$

1. Let $2.75\;\mathrm{cm}$ to the amplitude *A* of the wave.
2. Convert $0.410\;\mathrm{\frac{rad}{cm}}$ to $\mathrm{\frac{rad}{m}}$.
    $$
   \begin{align}
   0.410\;\mathrm{\frac{rad}{cm}} * \frac{100\;\mathrm{cm}}{\mathrm{m}} = 41.0\;\mathrm{\frac{rad}{cm}}
   \end{align}
    $$

4. Let $41.0\;\mathrm{\frac{rad}{m}}$ refer to the quantity $k x$, which is the wave number $k = \frac{2\pi}{\lambda}$ multiplied by the position *x*.
   $$
   \begin{align}
   41.0 \; \mathrm{\frac{rad}{m}} = k = \frac{2 \pi}{\lambda} \\
   41.0 \; \mathrm{\frac{rad}{m}} \times \frac{\lambda}{1} = \frac{2 \pi}{\lambda} \times \frac{\lambda}{1} \\
   41.0 \lambda \; \mathrm{\frac{rad}{m}} = 2 \pi \\
   \frac{41.0 \lambda}{41.0} \; \mathrm{\frac{rad}{m}} = \frac{2 \pi}{41.0} \\
   \lambda = \frac{2 \pi}{41.0} \mathrm{\frac{rad}{m}} \\
   \end{align}
   $$
   


```python
## Given
A = 2.75 * (1.0 / 100.0) # cm to m conversion
k = 0.410 * (100.0) ## rad/cm to rad/m
omega = 6.20

## lambda
lamb = math.tau / k
lamb
```

Thus $\lambda = 0.153 \text{ m}$.

5. Let $6.20 \mathrm{\frac{rad}{s}}$ refer the quantity $\omega t$, which is the angular frequency $\omega = 2\pi f = \frac{2\pi}{T}$ multiplied by the period *T*.
   $$
   \begin{align}
   6.20 \mathrm{\frac{rad}{s}} t = \omega t = 2 \pi f t \\
   \omega  = 6.20 \mathrm{\frac{rad}{s}} \\
   2 \pi f = 6.20 \mathrm{\frac{rad}{s}} \\
   \frac{2 \pi f}{2 \pi} = \frac{6.20}{2 \pi} \mathrm{\frac{rad}{s}} \\
         f = \frac{6.20 }{2 \pi} {\frac{1}{\text{s}}} \\
   \end{align}
   $$

```python
f = omega / math.tau
f
```

Thus the frequency $f = 0.987 \; \text{Hz}$.

1. How much time does it take for one complete wave pattern to go past a fisherman in a boat at anchor, and what horizontal distance does the wave crest travel in that time?


We know that the time period *T* is given by the formula $1 / f$. Thus:

$$
\begin{align}
T = \frac{1}{f} \text{ where } f = 0.987 \text{ Hz} \\
T = \frac{1}{0.987 \text{ Hz}} \\
\end{align}
$$

```python
T = 1.0 / f
T
```

Thus the time period $T = 1.01 \text{ s}$.

To calculate the horizontal distance, we use the wave number formula $k = \frac{2 \pi}{\lambda}$.

$$
\begin{align}
k = \frac{2 \pi}{\lambda} = \frac{2 \pi}{0.153 \text{ m}} \\
\end{align}
$$

2. What are the wave number and number of waves per second that pass the fisherman?


```python
v = lamb * f
(lamb, f, v)
```

```python
vmax = omega * A
(omega, A, vmax)
```
