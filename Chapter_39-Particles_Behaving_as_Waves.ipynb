## Louis de Broglie

French physicist in 1924 who propsed the bold idea that the particles can sometimes behave like waves. He said that a particle with mass $m$ moving at a speed $v$ has a speed $v$.

$$
\lambda = \frac{h}{p} = \frac{h}{mv}
$$

Where:

Lambda is the wavelength.
v is the particle speed.
m is the mass.

This formula works for particles moving at ordinary speeds, i.e. **not at all close to the speed of light**.
