2. An oscillator completes $12$ full cycles in $3.0 \; \text{s}$. What are its frequency $f$, period $t$, and angular frequency $\omega$?

We are given the following:
- Number of cycles $n = 12$
- Elapsed time $t = 3.0 \; \text{s}$

Recall that frequency is the number of cycles per second. So it can be represented by the following formula:

$$
\begin{align}
f &= \frac{n}{t} \\
  &= \frac{12}{3.0 \; \text{s}} \\
  &= 4.0 \; \text{Hz}
\end{align}
$$

We can then compute for period $T$ as it is the reciprocal of frequency $f$.

$$
\begin{align}
T &= \frac{1}{f} \\
  &= \frac{1}{4.0 \; \text{Hz}} \\
  &= 0.25 \; \text{s}
\end{align}
$$

Lastly, the angular frequency $\omega$ can be derived from another formula for frequency $f$.

$$
\begin{align}
(      f &= \frac{\omega}{2 \pi} ) \frac{2 \pi}{1} \\
  \omega &= 2 \pi f \\
         &= 2 \pi * 4.0 \; \text{Hz} \\
         &= 2 \pi * 4.0 \; \frac{1}{\text{s}} \\
         &= 8 \pi \frac{1}{\text{s}} \\
         &= 25.13 \frac{\text{rad}}{\text{s}}
\end{align}
$$

3. The amplitude $A$ of a mass-spring oscillator is doubled (with the same mass $m$ and spring constant $k$). What happens to the period $T$, maximum speed $v_{max}$, and total energy $E$?

Recall that in the graphs, the period $T$ is independent of the amplitude $A$. However, since $v_{max} = \omega A$, it doubles. Since $E = \frac{1}{2} k A^2$, its exponential nature means a doubling of the amplitude $A$ results in a quadrupling of the total energy $E$.

4. A glider on a spring is momentarily at $x = -A$. Which description is correct?

The maximum amplitude $A$ is at either extreme. In this case, the negative extreme (i.e. $-A$) is the leftmost. At this point, the glider has a speed $v_x = 0$ because it is about to switch directions. However, no matter the xtreme, the restoring force **always points toward equilibrium**.

5. Several tuning forks have tines of different mass. Which produces the lowest pitch?

Recall the formula for frequency, given some mass $m$:

$$
\begin{align}
f = \frac{1}{2} \pi \sqrt{\frac{k}{m}}
\end{align}
$$

Because the mass $m$ is in the denominator, this means that mass is indirectly proportional to the resulting frequency $f$. Thus larger masses result in lower frequencies, assuming the same spring constant $k$.

6. For $x(t) = A \cos(\omega t + \phi)$, which phase angle $\phi$ describes an object that is at $x = 0$ and $t = 0$ and moving in the $+x$ direction?

Recall the formula for displacement as a function of time for simple harmonic motion, substituting in displacement $x = 0 \; \text{m}$ and time $t = 0 \; \text{s}$. We are effectively looking for the phase angle at the **starting point** of the graph.

$$
\begin{align}
x(t) &= A \cos(\omega t + \phi) \\
0    &= A \cos(\omega \times 0 + \phi) \\
0    &= A \cos(\phi) \\
since \\
\cos(0) &= 1 \\
thus \\
0 &= A \cos(0) \\
0 &= A \times 1 \\
0 &= A
\end{align}
$$

8. A block with mass $m = 0.200 \; \text{kg}$, on a horizontal spring, with spring constant $k = 80.0 \; \frac{\text{N}}{\text{m}}$ oscillates with an amplitude $A = 0.100 \; \text{m}$. What are its maximum speed and maximum acceleration?

Recall the formulas used for *Energy in Simple Harmonic Motion*. Recall that $v_{max}$ occurs as displacement $x = 0$. So let's use the formula for some speed $v$:

$$
\begin{align}
v_x &= \pm \; \sqrt{\frac{k}{m}}
       \times \sqrt{A^2 - x^2} \\
    &= \pm \; \sqrt{\frac{k}{m}}
       \times \sqrt{A^2 - 0^2} \\
    &= \pm \; \sqrt{\frac{k}{m}}
       \times \sqrt{A^2} \\
    &= \pm \; \sqrt{\frac{k}{m}} \times A \\
\text{where} \\
\omega &= \sqrt{\frac{k}{m}} \\
\text{thus} \\
v_{max} &= \omega A
\end{align}
$$

Recall that the above formula was derived from the equation for energy.

$$
\begin{align}
E &= \frac{1}{2} m {v_x}^2 + \frac{1}{2} k x^2 = \frac{1}{2} k A^2 \\
\\
( \frac{1}{2} m {v_x}^2 + \frac{1}{2} k x^2 &= \frac{1}{2} k A^2 ) - \frac{1}{2} k x^2 \\
  \frac{1}{2} m {v_x}^2 &= \frac{1}{2} A^2 - \frac{1}{2} k x^2 \\
( \frac{1}{2} m {v_x}^2 &= \frac{1}{2} (A^2 - k x^2) ) \frac{2}{1} \\
( m {v_x}^2 &= A^2 - k x^2 ) \frac{1}{m} \\
{v_x}^2 &= \frac{A^2 - k x^2}{m} \\
v_x &= \pm \; \sqrt{\frac{A^2 - k x^2}{m}} \\
v_x &= \pm \; \frac{\sqrt{A^2 - k x^2}}{\sqrt{m}}
\end{align}
$$

So we can solve for the angular frequency $\omega$ first and then multiply it to the amplitude $A$ to get the maximum speed $v_{max}$.

$$
\begin{align*}
\omega &= \sqrt{\frac{k}{m}} \\
       &= \sqrt{\frac{80.0 \; \frac{\text{N}}{\text{m}}}
                     {0.200 \; \text{kg}}
               } \\
       &= \sqrt{\frac{80.0 \; \frac{\text{kg} \cdot \text{m} \cdot \text{s}^{-2}}
                                   {\text{m}}
                     }
                     {0.200 \; \text{kg}}
               } \\
       &= \sqrt{\frac{80.0 \; \text{s}^{-2}}
                     {0.200}
               } \\
       &= \sqrt{\frac{80.0 \; \frac{1}{\text{s}^2}}
                     {0.200}
               } \\
       &= \sqrt{400 \; \frac{1}{\text{s}^2}} \\
       &= 20 \; \frac{1}{\text{s}} \\
       &= 20 \; \frac{\text{rad}}{\text{s}}
\end{align*}
$$

Then:

$$
\begin{align*}
v_x &= \omega A \\
    &= 20 \; \frac{1}{\text{s}} \times 0.100 \; \text{m} \\
    &= 2.0 \; \frac{\text{m}}{\text{s}}
\end{align*}
$$

So the maximum speed is $v_{max} = 2.0 \; \frac{\text{m}}{\text{s}}$. To get the maximum acceleration $a_{max}$, we can use the following formula:

$$
\begin{align*}
a_{max} &= - \frac{k}{m} (-A) \\
        &= - \frac{80.0 \; \frac{\text{N}}{\text{m}}}
                  {0.200 \; \text{kg}}
             (- 0.100 \; \text{m}) \\
        &= - \frac{80.0 \; \text{kg} \cdot \text{m} \cdot \text{s}^{-2} \times \frac{1}{\text{m}}}
                  {0.200 \; \text{kg}}
             (- 0.100 \; \text{m}) \\
        &= - \frac{80.0 \; \text{s}^{-2}}
                  {0.200} \; 
             (- 0.100 \; \text{m}) \\
        &= 40 \; \text{m} \cdot \text{s}^{-2} \\
        &= 40 \; \frac{\text{m}}{\text{s}^2}
\end{align*}
$$

10. A block with mass $m = 0.50 \; \text{kg}$ on a spring oscillates at frequency $f = 2.0 \; \text{Hz}$. If it is replaced by another block with new mass $m' = 2.0 \; \text{kg}$ on the same spring, what is the new frequency?

First, we need to somehow relate mass and frequency. Recall that for an *angular* frequency $\omega$:

$$
\begin{align}
\omega &= \sqrt{\frac{k}{m}} \\
     f &= \frac{\omega}{2 \pi} = \frac{1}{2 \pi} \times \sqrt{\frac{k}{m}} \\
     T &= \frac{1}{f} = \frac{2 \pi}{\omega}
\end{align}
$$

The relation $T = \frac{1}{f}$ is useful because we can directly solve for the time period $T$ and relate it back to get the angular frequency $\omega$.

$$
\begin{align}
T &= \frac{1}{f} \\
  &= \frac{1}{2.0 \; \cfrac{1}{\text{s}}} = \frac{1}{2.0} \; \text{s} \\
  &= 0.5 \; \text{s}
\\
                              T &= \frac{2 \pi}{\omega} \\
              ( 0.5 \; \text{s} &= \frac{2 \pi}{\omega} ) \frac{\omega}{1} \\
( \omega \times 0.5 \; \text{s} &= 2 \pi ) \frac{1}{0.5 \; \text{s}} \\
                         \omega &= \frac{2 \pi}{0.5 \; \text{s}} \\
                                &= 12.57 \; \frac{1}{\text{s}} \\
                                &= 12.57 \; \frac{\text{rad}}{\text{s}}
\end{align}
$$

We can then use the angular frequency $f$ to calculate the spring force constant $k$. The $k$ for the initial block should be the same as the $k$ for the second one.

$$
\begin{align}
\omega &= \sqrt{\frac{k}{m}} \\
  12.57 \; \frac{\text{rad}}{\text{s}} &= \sqrt{\frac{k}{0.50 \; \text{kg}}} \\
( 12.57 \; \frac{\text{rad}}
                {\text{s}}
 &= \frac{\sqrt{k}}
         {\sqrt{0.50 \; \text{kg}} }
) \frac{\sqrt{0.50 \; \text{kg}} }
       {1} \\
( \sqrt{k} &= 12.57 \; \frac{\text{rad}}{\text{s}} \times \sqrt{0.50 \; \text{kg}} )^2 \\
k &= 157.0049 \; \frac{1}{\text{s}^2} \times 0.50 \; \text{kg} \\
k &= 79 \; \text{kg} \cdot \text{s}^{-2}
\end{align}
$$

We can now solve for the new frequency $f'$ when a block with some other mass $m' = 2.0 \; \text{kg}$.

$$
\begin{align}
f' &= \frac{1}{2 \pi} \times \sqrt{\frac{k}{m'}} \\
   &= \frac{1}{2 \pi}
      \times
      \sqrt{\frac{79  \; \text{kg} \cdot \text{m} \cdot \text{s}^{-2}}
                 {2.0 \; \text{kg}}
           } \\
   &= \frac{1}{2 \pi} \times \sqrt{39.5 \; \frac{\text{m}}{\text{s}^2}} \\
   &= 1 \; \text{Hz}
\end{align}
$$

13. A spring has spring force constant $k = 200 \;\frac{\text{N}}{\text{m}}$ and oscillates with amplitude $A = 0.050 \; \text{m}$. What is the kinetic energy at displacement $x = 0.030 \; \text{m}$?

The techanical energy $E$ is composed of kinetic energy $K$ and potential energy $U$. They are given by the following formulas:

$$
\begin{align}
E &= K + U \\
\text{where:} \\
K &= \frac{1}{2} m {v_x}^2 \\
U &= \frac{1}{2} k x^2 \\
\text{thus:} \\
E &= \frac{1}{2} m {v_x}^2 + \frac{1}{2} k x^2 \\
  &= \frac{1}{2} k A^2
\end{align}
$$

We can isolate the kinetic energy $K$.

$$
\begin{align}
K + U &= \frac{1}{2} k A^2 \\
K     &= \frac{1}{2} k A^2 - U \\
      &= \frac{1}{2} k A^2 - \frac{1}{2} k x^2 \\
      &= \frac{1}{2} (200 \; \frac{\text{N}}{\text{m}}
                      \times (0.050 \; \text{m})^2
                     )
         - \frac{1}{2} (200 \; \frac{\text{N}}{\text{m}}
                       \times (0.030 \; \text{m})^2
                       ) \\
      &= \frac{1}{2} (200 \; \frac{\text{N}}{\text{m}}
                     \times 0.0025 \; \text{m}^2)
         - \frac{1}{2} (200 \; \frac{\text{N}}{\text{m}}
                        \times 0.0009 \; \text{m}^2)  \\
      &= 0.5 \; \text{N} \cdot \text{m} - 0.18 \; \text{N} \cdot \text{m} \\
\\
K + U &= \frac{1}{2} k A^2 \\
K     &= \frac{1}{2} k A^2 - U \\
      &= \frac{1}{2} k A^2 - \frac{1}{2} k x^2 \\
      &= \frac{1}{2} k \times (A^2 - x^2) \\
      &= \frac{1}{2} (200) \frac{\text{N}}{\text{m}} \times ((0.050 \; \text{m})^2 - (0.030 \; \text{m})^2) \\
      &= 100 \frac{\text{N}}{\text{m}} \times 0.0016 \; \text{m}^2 \\
      &= 0.16 \; \text{N} \cdot \text{m} \\
      &= 0.16 \; \text{J}
\end{align}
$$

Thus, at displacment $x = 0.030 \; \text{m}$, the kinetic energy is $K = 0.16 \; \text{J}$.
